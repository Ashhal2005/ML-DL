```python
import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import DataLoader
from torchvision import datasets, transforms
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))]
)
```

```python
train_data = datasets.MNIST(root="./data", train=True, transform=transform, download=True)
test_data = datasets.MNIST(root="./data", train=False, transform=transform, download=True)
```

```python
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)
```

```python
train_data.data.shape
```

```python
class mnistModel(nn.Module):
  def __init__(self, ActFunc):
    super().__init__()

    self.flatten = nn.Flatten()

    self.network = nn.Sequential(
        nn.Linear(784,128),
        ActFunc,
        nn.Linear(128,64),
        ActFunc,
        nn.Linear(64,10)
    )

  def forward(self,x):
    x = self.flatten(x)
    return self.network(x)
```

```python
def train_model(model, train_loader, criterion, optimizer, epoch= 10):
  model.train()
  for epoch in range (epoch):
    total_loss = 0
    for img, label in train_loader:

      img = img.to(device)
      label = label.to(device)

      pred = model(img)
      loss = criterion(pred, label)
      optimizer.zero_grad()
      loss.backward()
      optimizer.step()

      total_loss += loss.item()

    avg_loss = total_loss/len(train_loader)

    print (f"loss for epoch-{epoch + 1} = {loss}  || Average loss {avg_loss:.4f}")
```

```python
def Eval (model, test_loader):
  model.eval()
  correct = 0
  total = 0

  with torch.no_grad():
    for image, label in test_loader:

      image = image.to(device)
      label = label.to(device)

      output = model (image)
      pred = torch.argmax(output , dim = 1)

      total += label.size(0)
      correct += (pred == label).sum().item()

      accuracy = (100*correct) / total
  return accuracy

```

```python
criterion = nn.CrossEntropyLoss()
```

```python

relu_model = mnistModel(nn.ReLU()).to(device)

optimizer = optim.Adam(relu_model.parameters(), lr=0.001)

train_model(relu_model,train_loader,criterion,optimizer)

Eval(relu_model, test_loader)
```

```python
sigmoid_model = mnistModel(nn.Sigmoid()).to(device)
sig_optimizer = optim.Adam(sigmoid_model.parameters(), lr = 0.001)
train_model(sigmoid_model, train_loader,criterion, sig_optimizer)

Eval(sigmoid_model, test_loader)
```

```python

tanh_model = mnistModel(nn.Tanh()).to(device)
tanh_optimizer = optim.Adam(tanh_model.parameters(), lr = 0.001)
train_model(sigmoid_model, train_loader,criterion, tanh_optimizer)

Eval(sigmoid_model, test_loader)
```
